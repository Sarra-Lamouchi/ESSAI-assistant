```python
%load_ext autoreload
%autoreload 2

import sys
sys.path.append("../src")

from chunker import load_final_text, chunk_document_custom, filter_chunks, CHUNKING_STRATEGY
from indexer import get_chroma_collection, index_chunks
import json
import os

all_chunks = []
for filename in CHUNKING_STRATEGY.keys():
    document = load_final_text(filename)
    chunks = chunk_document_custom(document)
    all_chunks.extend(chunks)
    print(f"{filename:<45} → {chunks[0]['chunk_method']:<25} → {len(chunks)} chunks")

print(f"\nTotal brut : {len(all_chunks)} chunks")

kept_chunks, rejected_chunks = filter_chunks(all_chunks, min_chars=100, filter_tables=True)
print(f"Conservés : {len(kept_chunks)} | Rejetés : {len(rejected_chunks)}")

output_path = "../data/processed/chunks/custom_chunks.jsonl"
os.makedirs(os.path.dirname(output_path), exist_ok=True)
with open(output_path, "w", encoding="utf-8") as f:
    for chunk in kept_chunks:
        f.write(json.dumps(chunk, ensure_ascii=False) + "\n")
print(f"✅ Sauvegardé dans {output_path}")

collection = get_chroma_collection(collection_name="essai_custom", force_recreate=True, mode="passage")
n_indexed = index_chunks(kept_chunks, collection, categorie="administratif")
print(f"✅ {n_indexed} chunks indexés")
```
